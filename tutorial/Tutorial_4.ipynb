## Part 4: Data Repair

With data loaded, denial constraints written, and error detection performed, we are now ready to perform actual repairs

### Data Loading

We'll quickly load in the information needed for repairs. Nothing new here if you've read our other tutorials

```python
from holoclean.holoclean import HoloClean, Session

# instantiate `HoloClean` and `Session` objects
holo = HoloClean(mysql_driver = "../holoclean/lib/mysql-connector-java-5.1.44-bin.jar" )
session = Session(holo)

# load data itself
data_path = "data/hospital_dataset.csv"
data = session.load_data(data_path)

# load denial constraints
dc_path = "data/hospital_constraints.txt"
dcs = session.load_denial_constraints(dc_path)

# load clean data defined by error detection
clean_path = "data/clean.csv"
clean = session.load_clean_data(clean_path)

# load dirty data defined by error detection
dirty_path = "data/dirty.csv"
dirty = session.load_dirty_data(dirty_path)
```
